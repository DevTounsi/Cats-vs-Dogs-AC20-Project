```python
#import 
import os #j'importe os pour pouvoir manipuler mes dossier et fichiers de data
from PIL import Image #j'importe Image pour pouvoir manipuler mes images

#Fonction permettant de detecter les images corrompues, les supprimer et me donner la hauteur et largeur moyenne
def analyse_data(data_folder):

    #variable
    tot_hauteur = 0 #variable pour stocker la somme des hauteurs des images
    tot_largeur = 0 #variable pour stocker la somme des largeurs des images
    hauteur_moyenne = 0
    largeur_moyenne = 0
    corrompuees = 0 #variable pour stocker le nombre d'images corrompues


    for dos in os.listdir(data_folder): 
        path_dos = os.path.join(data_folder, dos) #chemin du dossier dans data
        if os.path.isdir(path_dos): 
            print(f"Dossier : {dos}")
            for file in os.listdir(path_dos): #pour chaque fichier dans le dossier
                path_file = os.path.join(path_dos, file) #je crée une variable path_file qui contient le chemin du fichier
                if os.path.isfile(path_file):  
                    try: 
                        with Image.open(path_file) as img: #j'ouvre l'image avec Image.open et je la stocke dans la variable img
                            tot_hauteur += img.size[1] #j'ajoute la hauteur de l'image à la variable tot_hauteur
                            tot_largeur += img.size[0] #j'ajoute la largeur de l'image à la variable tot_largeur
                    except Exception as e: 
                        print(f"Erreur lors de l'ouverture de l'image {file} : {e}") #j'affiche l'erreur
                        os.remove(path_file) #je supprime le fichier corrompu
                        corrompuees += 1 #j'incrémente le nombre d'images corrompues
    hauteur_moyenne = tot_hauteur /(len(os.listdir(data_folder)) * len(os.listdir(path_dos)) - corrompuees)
    largeur_moyenne = tot_largeur / (len(os.listdir(data_folder)) * len(os.listdir(path_dos)) - corrompuees)
    print(f"Nombre d'images corrompues : {corrompuees}")
    print(f"Hauteur moyenne des images : {hauteur_moyenne}") #j'affiche la hauteur moyenne des images
    print(f"Largeur moyenne des images : {largeur_moyenne}") #j'affiche la largeur moyenne des images
```

```python
analyse_data("../data") #j'appelle la fonction analyse_data en lui passant le chemin du dossier data
```
